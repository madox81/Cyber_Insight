```python
# ==========================================================
# 1. CONFIGURATION & MAPPINGS (EXPANDED)
# ==========================================================

# Expanded Dictionary to cover more techniques found in Blue/Purple datasets

ID_TO_NAME = {
    # --- INITIAL ACCESS ---
    "T1190": "Exploit Public-Facing Application",
    "T1199": "Trusted Relationship",
    "T1566": "Phishing",
    "T1566.001": "Spearphishing Attachment",
    "T1566.002": "Spearphishing Link",
    "T1189": "Drive-by Compromise",
    "T1195.002": "Supply Chain Compromise: Software Dependencies",

    # --- EXECUTION ---
    "T1059": "Command and Scripting Interpreter",
    "T1059.001": "PowerShell",
    "T1059.003": "Windows Command Shell",
    "T1059.005": "Visual Basic",
    "T1059.006": "Python",
    "T1059.007": "JavaScript",
    "T1059.008": "Network Device CLI",
    "T1203": "Exploitation for Client Execution",
    "T1053": "Scheduled Task/Job",
    "T1053.005": "Scheduled Task/Job: Scheduled Task",
    "T1047": "Windows Management Instrumentation (WMI)",
    "T1204.001": "User Execution: Malicious Link",
    "T1204.002": "User Execution: Malicious File",

    # --- PERSISTENCE ---
    "T1547": "Boot or Logon Autostart Execution",
    "T1547.001": "Registry Run Keys / Startup Folder",
    "T1197": "BITS Jobs",
    "T1505": "Server Software Component",
    "T1505.003": "Web Shell",
    "T1505.004": "IIS Components",
    "T1543.003": "Create or Modify System Process: Windows Service",
    "T1176": "Browser Extensions",
    "T1546.003": "Event Triggered Execution: Windows Management Instrumentation Event Subscription",
    "T1546.012": "Event Triggered Execution: Image File Execution Options Injection",

    # --- PRIVILEGE ESCALATION ---
    "T1068": "Exploitation for Privilege Escalation",
    "T1134": "Access Token Manipulation",
    "T1134.001": "Token Impersonation/Theft",
    "T1548.005": "Abuse Elevation Control Mechanism: Setuid and Setgid",
    "T1078": "Valid Accounts",
    "T1611": "Escape to Host",

    # --- DEFENSE EVASION ---
    "T1562": "Impair Defenses",
    "T1562.001": "Disable or Modify Tools",
    "T1562.008": "Impair Defenses: Disable Cloud Logs",
    "T1070": "Indicator Removal",
    "T1070.001": "Clear Windows Event Logs",
    "T1070.004": "File Deletion",
    "T1112": "Modify Registry",
    "T1218": "System Binary Proxy Execution",
    "T1218.001": "Compiled HTML File",
    "T1218.004": "InstallUtil",
    "T1218.005": "Mshta",
    "T1218.010": "Regsvr32",
    "T1036": "Masquerading",
    "T1036.003": "Rename System Utilities",
    "T1055": "Process Injection",
    "T1055.001": "Dynamic-link Library Injection",
    "T1055.012": "Process Hollowing",
    "T1027": "Obfuscated Files or Information",
    "T1027.002": "Software Packing",
    "T1014": "Rootkit",
    "T1564.001": "Hide Artifacts: Hidden Files and Directories",
    "T1564.004": "Hide Artifacts: NTFS File Attributes",
    "T1553.002": "Subvert Trust Controls: Code Signing",
    "T1574": "Hijack Execution Flow",
    "T1574.002": "DLL Side-Loading",

    # --- CREDENTIAL ACCESS ---
    "T1003": "OS Credential Dumping",
    "T1003.001": "LSASS Memory",
    "T1003.002": "Security Account Manager",
    "T1558": "Steal or Forge Kerberos Tickets",
    "T1558.001": "Golden Ticket",
    "T1558.002": "Silver Ticket",
    "T1110": "Brute Force",
    "T1110.004": "Credential Stuffing",
    "T1552.005": "Unsecured Credentials: Cloud Instance Metadata Service",
    "T1552.007": "Unsecured Credentials: Container Service",
    "T1550.001": "Use Alternate Authentication Material: Application Access Token",
    "T1550.002": "Pass the Hash",
    "T1550.003": "Pass the Ticket",
    "T1056.001": "Keylogging",
    "T1056.002": "GUI Input Capture",

    # --- DISCOVERY ---
    "T1087": "Account Discovery",
    "T1087.001": "Local Account Discovery",
    "T1087.004": "Cloud Account Discovery",
    "T1018": "Remote System Discovery",
    "T1069": "Permission Groups Discovery",
    "T1069.001": "Local Groups",
    "T1069.002": "Domain Groups",
    "T1082": "System Information Discovery",
    "T1016": "System Network Configuration Discovery",
    "T1016.001": "Internet Connection Discovery",
    "T1033": "System Owner/User Discovery",
    "T1049": "System Network Connections Discovery",
    "T1135": "Network Share Discovery",
    "T1057": "Process Discovery",
    "T1518.001": "Software Discovery: Security Software",
    "T1012": "Query Registry",
    "T1007": "System Service Discovery",
    "T1083": "File and Directory Discovery",
    "T1046": "Network Service Discovery",
    "T1526": "Cloud Service Discovery",

    # --- LATERAL MOVEMENT ---
    "T1021": "Remote Services",
    "T1021.001": "Remote Desktop Protocol",
    "T1021.002": "SMB/Windows Admin Shares",
    "T1021.003": "Distributed Component Object Model (DCOM)",
    "T1021.004": "SSH",
    "T1021.006": "Windows Remote Management (WinRM)",
    "T1570": "Lateral Tool Transfer",
    "T1563.002": "Remote Service Session Hijacking: RDP Hijacking",
    "T1550.004": "Use Alternate Authentication Material: Web Session Cookie",

    # --- COLLECTION ---
    "T1560": "Archive Collected Data",
    "T1040": "Network Sniffing",

    # --- COMMAND AND CONTROL ---
    "T1071": "Application Layer Protocol",
    "T1071.001": "Web Protocols (HTTP/S)",
    "T1071.002": "File Transfer Protocols (FTP/SFTP)",
    "T1071.004": "DNS",
    "T1105": "Ingress Tool Transfer",
    "T1572": "Protocol Tunneling",
    "T1090": "Proxy",
    "T1090.003": "Proxy: Multi-hop Proxy",
    "T1573.001": "Encrypted Channel: Symmetric Cryptography",
    "T1573.002": "Encrypted Channel: Asymmetric Cryptography",
    "T1102": "Web Service",

    # --- EXFILTRATION ---
    "T1041": "Exfiltration Over C2 Channel",
    "T1048": "Exfiltration Over Alternative Protocol",
    "T1537": "Transfer Data to Cloud Account",

    # --- IMPACT ---
    "T1486": "Data Encrypted for Impact",
    "T1490": "Inhibit System Recovery",
    "T1485": "Data Destruction",
    "T1529": "System Shutdown/Reboot",
    "T1489": "Service Stop",
    "T1496": "Resource Hijacking",
    "T1498": "Network Denial of Service",
    "T1498.002": "Network DoS: Reflection Amplification",
    "T1484.001": "Domain Policy Modification: Group Policy Modification",
    "T1484.002": "Domain Policy Modification: Trust Policy Modification",
    "T1499": "Endpoint Denial of Service",
    "T1610": "Cloud Infrastructure Irremediable Destruction",
    "T1613": "Container and Cloud Storage Discovery",

    # --- RESOURCE DEVELOPMENT ---
    "T1584": "Compromise Infrastructure",
    "T1588.006": "Obtain Capabilities: Vulnerabilities",
}

# Map Technique IDs to their PRIMARY Tactic (to fill missing data in Blue Team)
TECHNIQUE_TO_TACTIC = {
    # Initial Access
    "T1190": "Initial Access",
    "T1199": "Initial Access",
    "T1566": "Initial Access",
    "T1566.001": "Initial Access",
    "T1566.002": "Initial Access",
    "T1189": "Initial Access",
    "T1078": "Initial Access",
    "T1195.002": "Initial Access",

    # Execution
    "T1059": "Execution",
    "T1059.001": "Execution",
    "T1059.003": "Execution",
    "T1059.005": "Execution",
    "T1059.006": "Execution",
    "T1059.007": "Execution",
    "T1059.008": "Execution",
    "T1203": "Execution",
    "T1053": "Execution",
    "T1053.005": "Execution",
    "T1047": "Execution",
    "T1204.001": "Execution",
    "T1204.002": "Execution",

    # Persistence
    "T1547": "Persistence",
    "T1547.001": "Persistence",
    "T1197": "Persistence",
    "T1505": "Persistence",
    "T1505.003": "Persistence",
    "T1505.004": "Persistence",
    "T1098": "Persistence",
    "T1543.003": "Persistence",
    "T1176": "Persistence",
    "T1546.003": "Persistence",
    "T1546.012": "Persistence",

    # Privilege Escalation
    "T1068": "Privilege Escalation",
    "T1134": "Privilege Escalation",
    "T1134.001": "Privilege Escalation",
    "T1548.005": "Privilege Escalation",
    "T1611": "Privilege Escalation",

    # Defense Evasion
    "T1562": "Defense Evasion",
    "T1562.001": "Defense Evasion",
    "T1562.008": "Defense Evasion",
    "T1070": "Defense Evasion",
    "T1070.001": "Defense Evasion",
    "T1070.004": "Defense Evasion",
    "T1112": "Defense Evasion",
    "T1218": "Defense Evasion",
    "T1218.001": "Defense Evasion",
    "T1218.004": "Defense Evasion",
    "T1218.005": "Defense Evasion",
    "T1218.010": "Defense Evasion",
    "T1036": "Defense Evasion",
    "T1036.003": "Defense Evasion",
    "T1055": "Defense Evasion",
    "T1055.001": "Defense Evasion",
    "T1055.012": "Defense Evasion",
    "T1027": "Defense Evasion",
    "T1027.002": "Defense Evasion",
    "T1014": "Defense Evasion",
    "T1564.001": "Defense Evasion",
    "T1564.004": "Defense Evasion",
    "T1553.002": "Defense Evasion",
    "T1574": "Defense Evasion",
    "T1574.002": "Defense Evasion",

    # Credential Access
    "T1003": "Credential Access",
    "T1003.001": "Credential Access",
    "T1003.002": "Credential Access",
    "T1558": "Credential Access",
    "T1558.001": "Credential Access",
    "T1558.002": "Credential Access",
    "T1110": "Credential Access",
    "T1110.004": "Credential Access",
    "T1552.005": "Credential Access",
    "T1552.007": "Credential Access",
    "T1550.001": "Credential Access",
    "T1550.002": "Credential Access",
    "T1550.003": "Credential Access",

    # Discovery
    "T1087": "Discovery",
    "T1087.001": "Discovery",
    "T1087.004": "Discovery",
    "T1018": "Discovery",
    "T1069": "Discovery",
    "T1069.001": "Discovery",
    "T1069.002": "Discovery",
    "T1082": "Discovery",
    "T1016": "Discovery",
    "T1016.001": "Discovery",
    "T1033": "Discovery",
    "T1049": "Discovery",
    "T1135": "Discovery",
    "T1057": "Discovery",
    "T1518.001": "Discovery",
    "T1012": "Discovery",
    "T1007": "Discovery",
    "T1083": "Discovery",
    "T1046": "Discovery",
    "T1526": "Discovery",
    "T1613": "Discovery",

    # Lateral Movement
    "T1021": "Lateral Movement",
    "T1021.001": "Lateral Movement",
    "T1021.002": "Lateral Movement",
    "T1021.003": "Lateral Movement",
    "T1021.004": "Lateral Movement",
    "T1021.006": "Lateral Movement",
    "T1570": "Lateral Movement",
    "T1563.002": "Lateral Movement",
    "T1550.004": "Lateral Movement",

    # Collection
    "T1560": "Collection",
    "T1040": "Collection",
    "T1056.001": "Collection",
    "T1056.002": "Collection",

    # Command and Control
    "T1071": "Command and Control",
    "T1071.001": "Command and Control",
    "T1071.002": "Command and Control",
    "T1071.004": "Command and Control",
    "T1105": "Command and Control",
    "T1572": "Command and Control",
    "T1090": "Command and Control",
    "T1090.003": "Command and Control",
    "T1573.001": "Command and Control",
    "T1573.002": "Command and Control",
    "T1102": "Command and Control",

    # Exfiltration
    "T1041": "Exfiltration",
    "T1048": "Exfiltration",
    "T1537": "Exfiltration",

    # Impact
    "T1486": "Impact",
    "T1490": "Impact",
    "T1485": "Impact",
    "T1529": "Impact",
    "T1489": "Impact",
    "T1496": "Impact",
    "T1498": "Impact",
    "T1498.002": "Impact",
    "T1499": "Impact",
    "T1484.001": "Impact",
    "T1484.002": "Impact",
    "T1610": "Impact",

    # Resource Development
    "T1584": "Resource Development",
    "T1588.006": "Resource Development",
}

# STRICT INSTRUCTIONS
MITRE_INSTRUCTIONS = [
    "Map the following security event to MITRE ATT&CK tactics and techniques.",
    "Identify the ATT&CK tactics and techniques in this data.",
    "Classify this cybersecurity event into MITRE ATT&CK framework.",
    "Extract the relevant tactics and techniques from the input."
]

SEVERITY_INSTRUCTIONS = [
    "Assess the severity and business risk of the following incident.",
    "Rate the severity (Low/Medium/High/Critical) and impact of this event.",
    "Provide a severity assessment for this security incident.",
    "Analyze the business risk and severity for the input below."
]
```

```python
# ==========================================================
# 2. HELPER FUNCTIONS
# ==========================================================

def get_technique_name(tech_id):
    """Returns Name or None if not found to enforce strict mapping."""
    return ID_TO_NAME.get(tech_id)

def create_sample(instruction, input_data, output_json, task_type):
    full_prompt = f"{instruction}\n\nInput: {input_data}"
    return {
        "messages": [
            {"role": "user", "content": full_prompt},
            {"role": "assistant", "content": output_json}
        ],
        "task_type": task_type
    }

def augment_task(input_data, output_json, task_type, n_variations=2):
    samples = []

    if task_type == "mitre":
        instruction_pool = MITRE_INSTRUCTIONS
    elif task_type == "severity":
        instruction_pool = SEVERITY_INSTRUCTIONS
    else:
        instruction_pool = ["Analyze the following:"]

    selected_instructions = random.sample(instruction_pool, min(n_variations, len(instruction_pool)))
    # Add standard version
    if task_type == "mitre":
        samples.append(create_sample(MITRE_INSTRUCTIONS[0], input_data, output_json, task_type))
    else:
        samples.append(create_sample(SEVERITY_INSTRUCTIONS[0], input_data, output_json, task_type))

    # Add variations
    for instr in selected_instructions:
        samples.append(create_sample(instr, input_data, output_json, task_type))

    return samples

def file_process(filepath):
  valid_records = []
  with open(filepath, 'r') as f:
    if filepath.endswith('.jsonl'):
      for line_num, line in enumerate(f, 1):
            try:
                record = json.loads(line)
                valid_records.append(record)
            except json.JSONDecodeError as e:
                print(f"Skipping malformed line {line_num} in {filepath}: {e} -> {line.strip()}")
            except Exception as e:
                print(f"An unexpected error occurred on line {line_num} in {filepath}: {e} -> {line.strip()}")
    else:
      valid_records = json.load(f)
  return valid_records

def oversample_severity(data, target_ratio=0.3):
    """Oversample severity samples to reach target ratio"""
    dataset = data.to_dict(orient='records')
    mitre_samples = [s for s in dataset if s['task_type'] == 'mitre']
    severity_samples = [s for s in dataset if s['task_type'] == 'severity']

    current_ratio = len(severity_samples) / len(dataset)

    if current_ratio < target_ratio:
        # Calculate how many duplicates needed
        target_severity = int(len(dataset) * target_ratio)
        repeats = target_severity // len(severity_samples)

        # Duplicate severity samples
        oversampled = severity_samples * repeats
        # Add remainder
        remainder = target_severity - len(oversampled)
        oversampled.extend(random.choices(severity_samples, k=remainder))

        # Combine with mitre samples
        balanced = mitre_samples + oversampled
        random.shuffle(balanced)

        return balanced

    return dataset
```

```python
# ==========================================================
# 3. PROCESSING FUNCTIONS (WITH STRICT FILTERING)
# ==========================================================

def process_blue_team(filepath):

    print(f"Processing Blue Team: {filepath}")
    data = file_process(filepath)
    df = pd.DataFrame(data)

    samples = []
    skipped = 0
    for _, row in df.iterrows():
        signature = str(row.get('signature', row.get('rule_type', '')))
        tech_id = str(row.get('mapped_technique', row.get('MITRE_ATT&CK_ID', '')))

        # 1. Get Tactic from row, OR infer it from ID
        tactic = str(row.get('tactic', '')).strip()
        if not tactic or tactic == 'nan':
            # If dataset lacks tactic, look it up in our dictionary
            tactic = TECHNIQUE_TO_TACTIC.get(tech_id, '')

        # 2. Skip only if we STILL don't have a tactic
        if not signature or not tech_id or not tactic:
            skipped += 1
            continue

        # 3. Enforce Name Mapping
        tech_name = get_technique_name(tech_id)
        if not tech_name:
            # Optional: Print ID if you want to see what's missing from dict
            print(f"Skipping unknown ID: {tech_id}")
            skipped += 1
            continue

        output = json.dumps({"tactics": [tactic], "techniques": [tech_name]})
        samples.extend(augment_task(signature, output, "mitre", n_variations = 5))

    print(f"Blue Team: Added {len(samples)} samples, skipped {skipped}.")
    return samples

def process_purple_team(filepath):

    print(f"Processing Purple Team: {filepath}")
    data = file_process(filepath)
    df = pd.DataFrame(data)

    samples = []
    skipped = 0
    for _, row in df.iterrows():
        desc = str(row.get('Event_Description', ''))
        tech_id = str(row.get('MITRE_ATT&CK_ID', ''))
        tactic = str(row.get('MITRE_Tactic', ''))

        if not desc or not tech_id or not tactic:
            skipped += 1
            continue

        # Enforce Name Mapping
        tech_name = get_technique_name(tech_id)
        if not tech_name:
            print(tech_id)
            skipped += 1
            continue

        output = json.dumps({"tactics": [tactic], "techniques": [tech_name]})
        samples.extend(augment_task(desc, output, "mitre", n_variations = 5))

    print(f"Purple Team: Added {len(samples)} samples, skipped {skipped}.")
    return samples

def process_playbook(filepath):
    print(f"Processing Playbook: {filepath}")
    data = file_process(filepath)

    if isinstance(data, dict) and 'incidents' in data:
        df = pd.DataFrame(data['incidents'])
    elif isinstance(data, list):
        df = pd.DataFrame(data)
    else:
        df = pd.DataFrame([data])

    samples = []
    skipped = 0
    for _, row in df.iterrows():
        inc_type = str(row.get('incident_type', ''))
        target = str(row.get('target_asset', ''))
        vector = str(row.get('initial_vector', ''))
        severity = str(row.get('severity', 'Medium'))

        tactics_techs = row.get('tactics_techniques', [])

        # --- TASK 1: MITRE MAPPING ---
        # Extract tactics and techniques
        tactics = list(set([item.get('tactic', '') for item in tactics_techs if item.get('tactic')]))
        techniques = list(set([item.get('technique', '') for item in tactics_techs if item.get('technique')]))

        # Skip if empty
        if not tactics or not techniques:
            skipped += 1
            continue

        input_mitre = f"Incident Type: {inc_type}\nTarget: {target}\nVector: {vector}"
        output_mitre = json.dumps({"tactics": tactics, "techniques": techniques})

        samples.extend(augment_task(input_mitre, output_mitre, "mitre", n_variations=3))

        # --- TASK 2: SEVERITY ASSESSMENT ---
        risk = "Security Breach"
        if "ransomware" in inc_type.lower(): risk = "Operational Shutdown"
        elif "phishing" in inc_type.lower(): risk = "Credential Compromise"

        input_sev = f"Incident: {inc_type} affecting {target}."
        output_sev = json.dumps({"severity": severity, "business_risk": risk})

        samples.extend(augment_task(input_sev, output_sev, "severity", n_variations=3))

    print(f"Playbook: Added {len(samples)} samples, skipped {skipped}.")
    return samples
```

```python
# ==========================================================
# 4. MAIN EXECUTION
# ==========================================================

if __name__ == "__main__":
    all_samples = []

    try:
        all_samples.extend(process_blue_team('BLUE_TEAM_DEFENSE_DATASET.jsonl'))
    except Exception as e:
        print(f"Skipping Blue Team: {e}")

    try:
        all_samples.extend(process_purple_team('purple_team_cybersecurity_dataset.json'))
    except Exception as e:
        print(f"Skipping Purple Team: {e}")

    try:
        all_samples.extend(process_playbook('incident_response_playbook_dataset.jsonl'))
    except Exception as e:
        print(f"Skipping Playbook: {e}")

    random.shuffle(all_samples)

    output_file = 'final_cyber_dataset.jsonl'
    with open(output_file, 'w') as f:
        for sample in all_samples:
            f.write(json.dumps(sample) + '\n')

    print(f"\n{'='*50}")
    print(f"✅ DATASET GENERATION COMPLETE")
    print(f"{'='*50}")
    print(f"Total Samples: {len(all_samples)}")
    print(f"Saved to: {output_file}")

    types = {}
    for s in all_samples:
        t = s.get('task_type', 'unknown')
        types[t] = types.get(t, 0) + 1
    print(f"Task Distribution: {types}")
```
